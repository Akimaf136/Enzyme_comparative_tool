```python
import urllib.request
import json
import ssl
import certifi

ssl_context = ssl.create_default_context(cafile=certifi.where())
#1. define the uniprot URL for the nonribosomal peptide synthetase enzyme

uniprot_ids = ["Q10896","Q6XME5"]

ssl_context = ssl.create_default_context(cafile=certifi.where())

all_data = {}

#2. open the web link and download the raw database file

for uniprot_id in uniprot_ids:
    url = f"https://rest.uniprot.org/uniprotkb/{uniprot_id}.json"

    response = urllib.request.urlopen(url, context=ssl_context)
    raw_data = json.loads(response.read())
    all_data[uniprot_id] = raw_data

    print(f"retrieved data for entry: {uniprot_id}")
```

```python
for uniprot_id, raw_data in all_data.items():
    #3. extraction the scientific name of the host organism
    organism_name = raw_data['organism']['scientificName']
    
    #4. extraction of the complete primary protein sequence
    protein_sequence = raw_data['sequence']['value']
    sequence_length = len(protein_sequence)
    
    print(f"Organism source: {organism_name}")
    print(f"Sequence length: {sequence_length} amino acids")
    print(f"First 10 Residues: {protein_sequence[:10]}...")
    print ()
```

```python
for uniprot_id, raw_data in all_data.items():
    # ensures the sequence for each Uniprot entry is extracted
    protein_sequence = raw_data['sequence']['value']
    sequence_length = len(protein_sequence)

    #5. count threonine residues as a measure of threonine composition
    threonine_count = protein_sequence.count("T")
    threonine_ratio = (threonine_count / sequence_length) * 100
    
    #6. count the hydrophobic core residues
    hydrophobic_residues = "AVILMFWY"
    hydrophobic_count = sum(1 for amino_acid in protein_sequence if amino_acid in hydrophobic_residues)
    hydrophobic_percentage = (hydrophobic_count / sequence_length) * 100
    
    print(f"UniProt ID: {uniprot_id}")
    print(f"threonine Content: {threonine_ratio:.2f}%")
    print(f"hydrophobic residue Content: {hydrophobic_percentage:.2f}%")
    print ()
```

```python
for uniprot_id, raw_data in all_data.items():
    print("===Candidate Screening===")
    print(f"(ID:{uniprot_id})")
    print(f"Production Host: {organism_name}")
    print(f"UniProt ID: {uniprot_id}")
    print(f"threonine Content: {threonine_ratio:.2f}%")
    print(f"hydrophobic residue Content: {hydrophobic_percentage:.2f}%")
    print ()
```
